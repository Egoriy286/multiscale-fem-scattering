# Локальные FEM-данные: рассеяние волн

Запускайте в локальном **Python (DOLFINx Complex)**. Все данные сохраняются в `neural_operator/data/low_frequency`; Google Drive и Colab не нужны. Первое выполнение создаёт данные, повторное использует каталог с той же конфигурацией.

На $\Omega=[-2,2]^2$ решаем для $u^s$:
\[
\int A\nabla u^s\cdot\nabla\bar v-k^2n u^s\bar v\,dx-ik\int_{\partial\Omega}u^s\bar v\,ds
=\int(1-A)\nabla u^i\cdot\nabla\bar v+k^2(n-1)u^i\bar v\,dx,
\quad u^i=e^{ik(x\cos\theta+y\sin\theta)}.
\]
Материалы DG0 задаются на общей конформной сетке с 24 кругами и крупными разделительными линиями. Одна LU-факторизация переиспользуется для направлений одной среды. Узловые A,n служат входами сети; для повторного FEM/GMsFEM сохраняются исходные поэлементные коэффициенты.


```python
from pathlib import Path
import sys
ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'neural_operator').is_dir())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
BASE = ROOT / 'neural_operator'
DATA_DIR = BASE / 'data' / 'low_frequency'
RUNS = BASE / 'runs' / 'low_frequency'
import json
import numpy as np
import matplotlib.pyplot as plt
from neural_operator.data import load_dataset
from neural_operator.plotting import display_table, plot_prediction, plot_history
```

```python
from neural_operator.fem import DatasetConfig, generate_dataset
CONFIG = DatasetConfig(n_media=48, n_directions=4, h=.10, n_coarse=4, k_range=(1., 2.), seed=2026)
dataset = generate_dataset(DATA_DIR, CONFIG)
print('Данные:', dataset.directory)
print('u_scat:', dataset.u_scat.shape, 'треугольников:', len(dataset.triangles))
```

```python
train, val, test = dataset.split_indices()
display_table([{'split': name, 'samples': len(ids), 'media': len(np.unique(dataset.medium_id[ids]))}
               for name, ids in zip(['train', 'validation', 'test'], [train, val, test])])
print('Интеграл единицы:', float(np.ones(len(dataset.coordinates)) @ dataset.mass_matrix @ np.ones(len(dataset.coordinates))))
```

```python
import matplotlib.tri as mtri
tri = mtri.Triangulation(*dataset.coordinates.T, dataset.triangles)
i = int(train[0])
fig, axes = plt.subplots(1, 3, figsize=(13, 4), constrained_layout=True)
for ax, values, title in zip(axes, [dataset.A[i], dataset.n[i], dataset.u_scat[i].real], ['A: nodal input', 'n: nodal input', 'Re scattered FEM field']):
    artist = ax.tripcolor(tri, values, shading='gouraud')
    ax.set(aspect='equal', title=title)
    fig.colorbar(artist, ax=ax, shrink=.8)
plt.show()
```
